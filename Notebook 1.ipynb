This block of code converts several columns in the global\_df dataset from float64 to int64 using the "\.astype\(np\.int64\)" method\. Originally, the columns like safety, education, health, and others were stored as floating\-point numbers, meaning they had decimal points\. Since these scores represent whole\-number rankings rather than continuous values, it made sense to convert them into integers\. This change helps make the data cleaner, more accurate for interpretation, and better suited for analysis and merging later on\.

```
#change columns safety and security, personal freedom, governance, social capital, investment environment, enterprise conditions, infrastructure, economic quality, living conditions,health, education, andnatural environment from float64 to integer

global_df['safety_and_security'] = global_df['safety_and_security'].astype(np.int64)
global_df['personal_freedom'] = global_df['personal_freedom'].astype(np.int64)
global_df['governance'] = global_df['governance'].astype(np.int64)
global_df['social_capital'] = global_df['social_capital'].astype(np.int64)
global_df['investment_environment'] = global_df['investment_environment'].astype(np.int64)
global_df['enterprise_conditions'] = global_df['enterprise_conditions'].astype(np.int64)
global_df['infrastructure_and_market_access'] = global_df['infrastructure_and_market_access'].astype(np.int64)
global_df['economic_quality'] = global_df['economic_quality'].astype(np.int64)
global_df['living_conditions'] = global_df['living_conditions'].astype(np.int64)
global_df['health'] = global_df['health'].astype(np.int64)
global_df['education'] = global_df['education'].astype(np.int64)
global_df['natural_environment'] = global_df['natural_environment'].astype(np.int64)

global_df
```

 We imported the GDP dataset using pd\.read\_csv\(\) and stored it in a DataFrame called gdp\_df\.
```
gdp_df = pd.read_csv("gdp_data.csv")
gdp_df
```

This code checks how many country names in the 'global\_df' dataset also exist in the 'gdp\_df' dataset\. The '\.isin\(\)' function creates a Boolean Series showing whether each country is present in both datasets\. By using '\.sum\(\)', we count how many of those values are 'True', which gives us the total number of exact matches between the two country lists\. This step is important for understanding how well the datasets align before merging\.

```
matching_countries = global_df['country'].isin(gdp_df['Country Name'])
exact_matches = matching_countries.sum()
exact_matches
```

To identify countries in global\_df that had no match in the GDP dataset, we calculated the difference between the total number of countries and the number of exact matches using len\(global\_df\) \- exact\_matches\. A new DataFrame, mismatched\_countries\_df, was then created to display only the unmatched countries\. This made it easy to locate and correct country name discrepancies before performing the final merge\.

```
mismatched_countries = len(global_df) - exact_matches
mismatched_countries

# create a list of mismatched countries
mismatched_countries_df = global_df[~matching_countries]
mismatched_countries_df
```

We converted the 'Country Name' column from the gdp\_df dataset into a Python list using \.tolist\(\)\. Here, a dictionary was created which we called manual\_corrections to fix differences in country names between the two datasets\. Some country names in global\_df did not exactly match the names in the GDP dataset, so we mapped them manually to their correct forms\. We then used \.apply\(\) with a lambda function to create a new column called Countries, applying the corrections where needed\. This was important to ensure that the country names aligned properly for a successful merge later on\.

```

manual_corrections = {
    'Hong Kong': 'Hong Kong SAR, China',
    'South Korea': 'Korea, Rep.',
    'Slovakia': 'Slovak Republic',
    'Vietnam': 'Viet Nam',
    'Russia': 'Russian Federation',
    'Kyrgyzstan': 'Kyrgyz Republic',
    'Turkey': 'Turkiye',
    'Laos': 'Lao PDR',
    'Egypt': 'Egypt, Arab Rep.',
    'Iran': 'Iran, Islamic Rep.',
    'Venezuela': 'Venezuela, RB',
    'Congo': 'Congo, Rep.',
    'Syria': 'Syrian Arab Republic',
    'Yemen': 'Yemen, Rep.'
}

global_df['Countries'] = global_df['country'].apply(lambda x: manual_corrections.get(x, x))

global_df
```

To determine how many country names in global\_df matched those in gdp\_df, we utilized the \.isin\(\) function\. This method returned a boolean Series indicating whether each country in global\_df was present in gdp\_df\. By summing this Series using \.sum\(\), we obtained the total count of exact matches\. This process was crucial for assessing the compatibility of the datasets and ensuring a successful merge\.

```
matching_countries = global_df['Countries'].isin(gdp_df['Country Name'])
exact_matches = matching_countries.sum()
exact_matches
```

In this step, we performed an inner merge between the 'gdp\_df' and 'global\_df' datasets using the 'merge\(\)' function\. We matched the 'Country Name' column from the GDP dataset with the 'Countries' column from the prosperity dataset\. An inner join was used so that only countries appearing in both datasets would be included in the final merged DataFrame, 'df\_merged'\. This ensures the resulting data contains complete and consistent information for each country included in the analysis\.

```
df_merged = gdp_df.merge(global_df, how = "inner", left_on = "Country Name", right_on = "Countries" )
df_merged
```

 We renamed two columns in the merged dataset for clarity and consistency\. We changed 'average\_score' to 'prosperity\_score' to better reflect what the column represents, and '2023' to '2023\_gdp' to clearly indicate that the values are GDP figures for the year 2023\. Renaming helps make the dataset easier to read and interpret during analysis\.

```
#rename average_score to prosperity_score and 2023 to 2023_gdp

df_merged.rename(columns={'average_score': 'prosperity_score', '2023': '2023_gdp'}, inplace=True)
df_merged
```

In this step, we removed any rows from the 'df\_merged' dataset where the '2023\_gdp' column contained missing \(null\) values using the 'dropna\(\)' function\. This was important to ensure that every country in our analysis had a valid GDP value for 2023, helping us maintain data accuracy and avoid errors in any calculations or visualizations that rely on this column\.

```
df_merged = df_merged.dropna(subset=['2023_gdp'])
df_merged
```

This code counts how many times each unique value appears in the 'political\_regime' column of the 'df\_merged' dataset using '\.value\_counts\(\)'\. It helps us understand the distribution of different political regimes across the countries in our merged dataset, which is useful for identifying patterns or imbalances in the data\.

```
df_merged['political_regime'].value_counts()
```

In this step, we created a new column called 'Regime\_Group' in the merged dataset to classify countries based on their political systems\. First, we made a copy of the DataFrame to avoid modifying the original data\. Then, using a custom function with '\.apply\(\)', we grouped countries into either "Democracy" or "Autocracy" depending on their \`political\_regime\` values\. This categorization helps simplify analysis by grouping similar regime types under broader labels\.

```
df_merged = df_merged.copy()

# Create Regime_Group
def regime_group(row):
    if row['political_regime'] in ['Electoral democracy', 'Liberal democracy']:
        return 'Democracy'
    elif row['political_regime'] in ['Electoral autocracy', 'Closed autocracy']:
        return 'Autocracy'
    else:
        return None

df_merged['Regime_Group'] = df_merged.apply(regime_group, axis=1)
```

Viewing our merged dataset

```
df_merged
```

This code checks how many countries in the merged dataset fall under each regime type by counting the values in the 'Regime\_Group' column\. In this context, '1' represents democracies and '0' represents autocracies\. The 'value\_counts\(\)' function gives a quick summary of how many countries belong to each group, helping us understand the political distribution in the dataset\.

```
# Check how many democracies and autocracies we have now
print(df_merged['Regime_Group'].value_counts())

# 1 = Democracy
# 0 = Autocracy
```

In this step, we renamed the '2023\_gdp' column in the merged dataset to 'gdp\_2023' for better readability and consistency in naming\. Additionally, we established a simple binary classification where '1' represents a democracy and '0' represents an autocracy, which may be used later for analysis or filtering based on political systems\.

```
df_merged = df_merged.rename(columns={'2023_gdp': 'gdp_2023'})
```

The 'statsmodels' library was used to build a linear regression model that predicts a country's gdp\_2023 based on factors such as infrastructure access, regime type, education, health, economic quality, living conditions, and social capital\. An interaction term between infrastructure\_and\_market\_access and Regime\_Group was included to examine how the impact of infrastructure on GDP may differ depending on the political regime\. The \.fit\(\) function runs the regression, while model\.summary\(\) displays detailed results, including coefficients, significance levels, and overall model performance\.

```
import statsmodels.formula.api as smf

model = smf.ols('gdp_2023 ~ infrastructure_and_market_access * Regime_Group + education + health + economic_quality + living_conditions + social_capital', data=df_merged).fit()
print(model.summary())
```

We used statsmodels and matplotlib to create an influence plot based on Cook’s Distance\. After fitting our model, we generated the plot using sm\.graphics\.influence\_plot\(\), which helps identify data points that have a large influence on the model’s results\. By visualizing the influence of each observation, we can detect outliers or influential points that might affect the stability and accuracy of our model\. The plot provides a clearer understanding of how individual data points impact the regression analysis\.

```
import statsmodels.api as sm
import matplotlib.pyplot as plt

# Assuming your model is already fitted
fig, ax = plt.subplots(figsize=(8,6))
sm.graphics.influence_plot(model, ax=ax, criterion="cooks")
plt.title('Influence Plot (Cook\'s Distance)')
plt.show()
```

This code helps remove two high\-influence points from the merged dataset using \.drop\(index=\[151, 25\]\) to create a cleaner version called df\_clean\. High\-influence points are data points that can heavily distort a regression model’s results\. After removing them, we refit the linear regression model using smf\.ols\(\), keeping the same variables and interaction terms as before\. Finally, we printed the new model summary to check if removing those points improved the model’s performance and stability\.

```
# Drop high-influence points
df_clean = df_merged.drop(index=[151, 25])

# Refit the model
model_clean = smf.ols('gdp_2023 ~ infrastructure_and_market_access * Regime_Group + education + health + economic_quality + living_conditions + social_capital', data=df_clean).fit()

# Print new summary
print(model_clean.summary())
```

Seaborn’s lmplot function was used to create a scatterplot with regression lines, illustrating the relationship between infrastructure access and GDP for 2023\. Data points were colored by Regime\_Group to visually compare democracies and autocracies, with distinct markers used to enhance clarity between the two\. This plot highlights how infrastructure correlates with GDP under different political systems and helps reveal whether the patterns vary across regime types\. Titles and labeled axes were added to improve readability and interpretation\.

```
import seaborn as sns
import matplotlib.pyplot as plt

# Plot
sns.lmplot(
    data=df_clean, 
    x='infrastructure_and_market_access', 
    y='gdp_2023', 
    hue='Regime_Group',         # Different colors for Democracy vs Autocracy
    markers=["o", "x"],          # Different markers
    palette="Set2",              # Nice color palette
    height=6,
    aspect=1.5
)

plt.title('Infrastructure vs GDP by Regime Type')
plt.xlabel('Infrastructure and Market Access')
plt.ylabel('GDP 2023')
plt.show()
```
